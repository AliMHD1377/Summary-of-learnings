انواع نمودارهای کلیدی seaborn:
</div>

- **Relational plots**: Seaborn provides functions like `scatterplot()`, `lineplot()`, and `relplot()` for visualizing the relationship between two or more variables. رابطه بین دو یا چند متغیر
- **Categorical plots**: Functions like `barplot()`, `countplot()`, `boxplot()`, `violinplot()`, and `stripplot()` allow you to visualize the distribution and summary statistics of categorical variables.
- **Distribution plots**: Seaborn offers functions like `histplot()`, `kdeplot()`, `rugplot()`, and `distplot()` for visualizing the distribution of univariate or bivariate data. توزیع دادها ی تک یا دو متغیره
- **Regression plots**: Functions like `regplot()` and `lmplot()` enable you to visualize linear relationships and perform regression analysis.
- **Matrix plots**: Seaborn provides `heatmap()` and `clustermap()` functions for visualizing matrix-like data and hierarchical clustering.
- **Multi-plot grids**: Functions like `catplot()`, `lmplot()`, `pairplot()`, and `jointplot()` allow you to create multi-plot grids to visualize relationships between multiple variables. روابط بین چندین متغیر

انواع نمودارهای seaborn :
</div>

1. **Relational**: For visualizing relationships between variables (e.g., `scatterplot`, `lineplot`)
2. **Distributional**: For representing univariate or bivariate distributions (e.g., `histplot`, `kdeplot`, `ecdfplot`)
3. **Categorical**: For plotting categorical data (e.g., `boxplot`, `violinplot`, `swarmplot`)
4. **Regression**: For visualizing linear relationships (e.g., `regplot`, `lmplot`)
5. **Matrix**: For visualizing matrix-structured data (e.g., `heatmap`, `clustermap`)
