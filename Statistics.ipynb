- Branches of Statistics  
  - Descriptive Statistics : نمونه را شرح میدهد
  - Inferential Statistics : نتیجه نمونه را به کل جامعه تعمیم میدیم

**Regression Analysis**: چه چیزی باعث این اثر شده. روشی برای مدل‌سازی رابطه بین یک متغیر وابسته (خروجی) و یک یا چند متغیر مستقل (ورودی) به منظور پیش‌بینی.

</div>

- independent = cause = مستقل - وقتی تغییرمیکنه اثری روی بقیه می گذارد
- dependent = effect = وابسته - مقدارش به متغیر دیگر بستگی دارد

### **02-05**
## Regression
<div dir='rtl'>

**پیش‌بینی** مقدار یک متغیر بر اساس متغیر دیگر. مدل‌سازی رابطه‌ی بین یک متغیر پاسخ (وابسته) و یک یا چند متغیر پیش‌بین (مستقل) — معمولاً برای پیش‌بینی یا تبیین.

دونوع از انواع رگرسیون :
</div>

1. **Simple Linear Regression** :
     - $y = \beta_0 + \beta_1x + \epsilon$
2. **Multiple Regression** :
     - $y = \beta_0 + \beta_1x_1 + \beta_2x_2 + \epsilon$
- هردو مدل به صورت خطی اند.
<div dir='rtl'>

**The Regression Line** :

در رگرسیون خطی ساده، رابطه میان متغیر مستقل (x) و متغیر وابسته (y) با استفاده از یک خط راست که «خط رگرسیون» نامیده می‌شود، مدل‌سازی می‌گردد. خط رگرسیون، خطی است که بهترین برازش(fit) را بر نقاط داده در یک نمودار پراکندگی (scatter plot) دارد و تفاوت میان مقادیر مشاهده‌شده و مقادیر پیش‌بینی‌شده توسط این خط را به حداقل می‌رساند.

**Least Squares Regression | کمترین توان 2ها** :
رایج‌ترین روش برای یافتن بهترین خط رگرسیون.

### **02-06**
## Evaluating the Accuracy of Regression Predictions | ارزیابی دقت پیش‌بینی‌های رگرسیون
 
**Standard Error of Estimate ($s_{y|x}$)** : داده های ما از اون چیزی که پیش بینی کردیم چه قدر فاصله داره(خط رگرسیون چه قدر خطا دارد)

<div dir='rtl'>

**The Coefficient of Determination ($r^2$) | ضریب پیش بینی(تعیین)** :

یک روش اندازه گیری آماری است که بیان میکند چه قدر از تغییرات y قابل پیشبینی توسط x است.(این خط رگرسیون چه قدر در پیش بینی خوب است؟)
- قدرت خط رگرسیون را نشان میدهد.
- میتونیم دوتا خط رگرسیون رو باهمین روش مقایسه کنیم.
- هرچه به یک نزدیکتر باشه خطا کمتر و هرچه به صفر نزدیکتر باشه خطای بیشتر
- محدودیت: حساس به داده پرت - فقط رابطه خطی

**مقایسه $r^2$ و $s_{y|x}$** :
- **$r^2$ (ضریب تعیین):** برای درک **قدرت رابطه** و مقایسه مدل‌های مختلف (بدون در نظر گرفتن واحد اندازه‌گیری متغیرها) استفاده می‌شود.
- **$s_{y|x}$ (خطای معیار):** برای درک **میزان خطای پیش‌بینی** در واحد اصلی داده‌ها و ارتباط دادن نتایج به افراد غیرمتخصص استفاده می‌شود.(هم واحد)(تصور راحت تری داره)
