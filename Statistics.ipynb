### **01-01**
<div dir='rtl'>

آمار: **آمار شاخه‌ای از ریاضیات است که به جمع‌آوری، تحلیل، تفسیر و ارائه داده‌ها می‌پردازد**. سپس یک نتیجه گیری میکنیم و در آخر یک تصمیمی میگیریم اونم در شرایطی که عدم قطئیت وجود دارد.

**هوش مصنوعی**: انواع و اقسام الگوریتم ها در دنیا وجود دارند اما یک سری از آنها مربوط به این هستند که هوش انسانی رو پیاده سازی بکنند.

**یادگیری ماشین**: ML هم می خواد هوش انسانی رو پیاده سازی بکنه اما فقط با استفاده از داده. برخلاف هوش مصنوعی که با هرروشی می خواد این کاررو انجام بده.

</div>

- Branches of Statistics  
  - Descriptive Statistics : نمونه را شرح میدهد
  - Inferential Statistics : نتیجه نمونه را به کل جامعه تعمیم میدیم
<div dir='rtl'>

- **Alternative Hypothesis (Ha) | فرضیه جایگزین**: شروع شک. وقتی میگی یک فرقی هست یا داره یک اتفاقی میفته.
- **Null Hypothesis (H0) | فرضیه اصلی**: هیچ شکی نیست. همه چی طبیعیه و هیچ اتفاقی نمیفته و هیچ فرقی نیست.

**Confidence Intervals | فواصل اطمینان**: آزمون فرضیه(Hypothesis Testing) صرفاً نشان می‌دهد که آیا اثری وجود دارد یا خیر. فاصله اطمینان، میگه حالا این چیزی که وجود دارد چه قدر است(شدتش چه قدره)

**Regression Analysis**: چه چیزی باعث این اثر شده. روشی برای مدل‌سازی رابطه بین یک متغیر وابسته (خروجی) و یک یا چند متغیر مستقل (ورودی) به منظور پیش‌بینی.

</div>

- independent = cause = مستقل - وقتی تغییرمیکنه اثری روی بقیه می گذارد
- dependent = effect = وابسته - مقدارش به متغیر دیگر بستگی دارد

### **02-01**
## Averages
<div dir='rtl'>

**Mode**: اونی که بیشتر از همه تکرار شده.(الان چی بپوشم مده؟)
- برای هرنوع داده ای قابل اعماله
- داده های پرت روش هیچ تاثیری نداره.

**Medain**: وسط
- برای همه داده ها به غیر از کیفی-اسمی(چون ترتیب ندارند)
- داده های پرت تاثیرکمی روش دارن

**Mean**: میانگین
- فقط برای داده های کمی
- بسیار حساس به داده های پرت
- حرف اختصاری میانگین برای نمونه $ \bar{x} $ اما برای کل جامعه $ \mu $ هستش.
