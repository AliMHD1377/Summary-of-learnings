<div dir='rtl'>

انواع یادگیری ماشین:
### Supervised Learning | یادگیری با نظارت
- خودمون داده رو lable می کنیم

- الگوریتم هایی که از این نوع اند:
</div>

1. Regression Algorithms:
   - Linear Regression
   - Polynomial Regression
   - Ridge Regression
   - Lasso Regression

2. Classification Algorithms:
   - Logistic Regression
   - Decision Trees
   - Random Forests
   - Support Vector Machines (SVM)
   - Naive Bayes
   - K-Nearest Neighbors (KNN)
<div dir='rtl'>

### Unsupervised Learning | یادگیری بدون نظارت
- طبق یک الگویی که خودش پیدا میکنه، داده هارو گروه بندی می کنه

- الگوریتم هایی که از این نوع اند:
</div>

1. Clustering Algorithms:
   - K-means Clustering
   - Hierarchical Clustering
   - DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

2. Dimensionality Reduction Algorithms:
   - Principal Component Analysis (PCA)
   - t-SNE (t-Distributed Stochastic Neighbor Embedding)
   - Autoencoders
<div dir='rtl'>

### Reinforcement Learning | یادگیری تقویتی
- یادگیری برحسب پاداش و جریمه ای که خودمون بهش میدیم.
- به نبال بیشترین پاداشه

- الگوریتم هایی که از این نوع اند:
</div>

1. Value-Based Methods:
   - Q-Learning
   - SARSA (State-Action-Reward-State-Action)
   - Deep Q-Networks (DQN)

2. Policy-Based Methods:
   - Policy Gradient
   - REINFORCE
   - Actor-Critic Methods

**02-02**
<div dir='rtl'>

overfitting: بیش اندازه به جزئیات نگاه کردن به جای اینکه به الگوی کلی توجه کنیم.

**تکنیک های انتخاب ویژگی:**

<img src="./images/feature-selection.png" width="600">

Filter Methods: براساس معیارهای آماری و مستقل از هرگونه الگوریتم یادگیری ماشین.

Wrapper Methods(پوششی): براساس الگوریتم یادگیری ماشین انتخاب میکنه.
- Backward Feature Elimination: همه ویژگی هارو انتخاب میکنیم و یکی یکی کم میکنیم تا ببینیم چی میشه.

Embedded Methods: در روش‌های تعبیه‌شده، انتخاب ویژگی و آموزش مدل به‌صورت هم‌زمان انجام می‌شوند. در این فرآیند، خودِ الگوریتم یادگیری ماشین مسئول تعیین میزان اهمیت یا مرتبط بودن هر ویژگی است. در طی پروسه آموزش، ما میتونیم بفهمیم کدوم ویژگی مهمه و کدوم مهم نیست.

- Lasso Regression (L1 Regularization): میزنیم توی سر همه ی ویژگی ها تا ضعیف ها بمیرند و قوی ها بمونند. پنالتی میزاره و همه کوچیک میشن ولی اونایی که خیلی کوچیکند از بین میرند.
</div>

**Feature Extraction Techniques:**    
  - Principal Component Analysis (PCA) 
  - Linear Discriminant Analysis (LDA)   
  - t-SNE (t-Distributed Stochastic Neighbor Embedding)   
  - Autoencoders
