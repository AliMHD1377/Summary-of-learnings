### **02-04**
## Common Discrete Probability Distributions | توزیع احتمالاتی گسسته:
1. Bernoulli Distribution : (با یک بار آزمایش)دوحالت بیشتر نداره یا موفق یا شکست
2. Binomial Distribution : تعداد بار موفقیت یا شکست توی چندبار آزمایش مستقل
3. Multinomial Distribution : (تعدادبار حالت خاص توی چندبار آزمایش مستقل)باینومیالی که تعداد حالت های ممکنش بیشتر از دوتاست
4. Poisson Distribution : تعداد حالات تصادفی و مستقل که در یک بازه زمانی یا مکانی ثابت رخ می دهد(در این بازه چند رویداد؟)
5. Geometric Distribution : چندبار تکرار کنم(آزمایش های برنولی مستقل) تا به اولین موفقیتم برسم؟(صبر تا پیروز بشی)
6. Negative Binomial Distribution
7. Hypergeometric Distribution

### **02-06**
## Frequentist vs Bayesian
<div dir='rtl'>

**Statistical Inference | استنباط آماری** : تصمیم گیری باتوجه به آماری که به دست آوردیم.

دو الگو اصلی استنتاج آماری:

1. **رویکرد Frequentist**: این رویکرد را مانند کارآگاهی در نظر بگیرید که دقیقاً طبق کتاب (و اصولِ سفت‌وسخت) عمل می‌کند و صرفاً بر شواهدِ موجود تکیه دارد.
2. **رویکرد بیزی (Bayesian)**: این رویکرد بیشتر شبیه کارآگاهی است که از شهود خود بهره می‌گیرد و تجربیات گذشته و حدس‌ و گمان‌ها را نیز در نظر می‌آورد.

مفاهیم کلیدی Frequentist:
- احتمال یعنی «اگه این آزمایش رو بی‌نهایت بار تکرار کنیم، چند درصد مواقع این نتیجه رو میبینیم؟>>(فرکانس بلند مدت)

مفاهیم کلیدی Bayesian:
- احتمال = درجه‌ی باور / معیار عدم‌قطعیت من درباره‌ی یک گزاره، با توجه به اطلاعات موجود، که با داده‌های جدید به‌روزرسانی می‌شود.
- بیزی‌ها احتمال را ابزاری برای بیان عدم‌قطعیت و باور می‌دانند، نه صرفاً یک فراوانی عینی و تکرارشونده.
- پارامترها رو به عنوان متغیر درنظر میگیره و سعی میکنه که اونهارو به دست بیاره.

تفاوت این دو:
- فراوانی‌گرا: حقیقت یک عدد ثابت است؛ من فقط آن را نمی‌دانم. عدم‌قطعیت در دانش من است.
- بیزی: حقیقت برای من یک توزیع احتمال دارد. عدم‌قطعیت در خودِ حقیقت است.

🎁 تشبیه ساده: جعبه‌ی بسته و عدد درون آن

فرض کن یک **جعبه‌ی بسته** داری که داخلش یک عدد نوشته شده. این عدد **حقیقت** است.

🔵 فراوانی‌گرا (Frequentist)

- می‌گوید: «داخل جعبه یک عدد **ثابت** نوشته شده. من نمی‌دانم چیست، ولی ثابت است.»
- با آزمایش‌ها، **حدس** می‌زند آن عدد چیست. عدم‌قطعیت درباره‌ی **حدس** است.

🟢 بیزی (Bayesian)

- می‌گوید: «داخل جعبه یک عدد است، ولی من **درباره‌ی خودِ آن عدد** نامطمئنم.»
- برای آن عدد یک **توزیع احتمال** در نظر می‌گیرد: «با احتمال ۳۰٪ عدد ۵ است، با احتمال ۲۰٪ عدد ۶ است، ...»
- بعد از دیدن شواهد، این توزیع را به‌روز می‌کند.

📌 خلاصه‌ی تشبیه

| مکتب | عدم‌قطعیت کجاست؟ |
|---|---|
| **فراوانی‌گرا** | در **حدس و دانش ما** درباره‌ی عدد |
| **بیزی** | در **خودِ عدد** (حقیقت) |

> **نکته:** هر دو مکتب می‌گویند «ما نامطمئن هستیم». تفاوت در این است که **عدم‌قطعیت را کجا می‌گذارند**:
> - فراوانی‌گرا: عدم‌قطعیت در **دانش ما**.
> - بیزی: عدم‌قطعیت در **خودِ حقیقت**.

### **02-07**
## Bayesian

$$ P(A|B) = \frac{P(B|A) \times P(A)}{P(B)} $$

**Prior - $P(A)$** : چیزی که قبل از دیدن داده می دونیم یا فکر میکنیم که میدونیم(باور اولیه)

**Likelihood - $P(B|A)$** : چه قدر احتمال داره که این مشاهده جدید دیده بشه به شرطی که فرضیه مون درست باشه. (شواهد. داده جدید)

**Posterior - $P(A|B)$** : باور بروز شده(هدف ما)

**Evidence - $P(B)$** : در محاسبه‌ی یک پوسترییر فقط نرمال‌ساز است، اما در مقایسه‌ی مدل‌ها نقش اصلی را دارد.
