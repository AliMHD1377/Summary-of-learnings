## Regression
<div dir='rtl'>

**پیش‌بینی** مقدار یک متغیر بر اساس متغیر دیگر. مدل‌سازی رابطه‌ی بین یک متغیر پاسخ (وابسته) و یک یا چند متغیر پیش‌بین (مستقل) — معمولاً برای پیش‌بینی یا تبیین.

در اصل، رگرسیون به ما این امکان را می‌دهد که:

1. مقدار یک متغیر وابسته را بر اساس مقادیر متغیرهای مستقل *پیش‌بینی* کنیم.
2. ماهیت و شدت روابط میان متغیرها را *تبیین* (یا تشریح) کنیم.
</div>

**Types of Regression Problems** :
- Simple vs. Multiple Regression
    - $y = f(x) + \epsilon$
    - $y = f(x_1, x_2, ..., x_n) + \epsilon$
- Linear vs. Non-linear Regression
    - $y = f(X, \beta) + \epsilon$, where $f$ is a non-linear function
- Other Types of Regression
    1. **Logistic Regression**: Used for binary classification problems, but technically a regression method.
    2. **Ridge Regression**: Linear regression with L2 regularization to prevent overfitting.
    3. **Lasso Regression**: Linear regression with L1 regularization, useful for feature selection.
    4. **Elastic Net**: Combines both L1 and L2 regularization.
    5. **Polynomial Regression**: Fits a non-linear relationship using polynomial terms.
    6. **Stepwise Regression**: Iteratively adds or removes predictors to find the best model.

**Key Concepts in Regression** :  
- Dependent and Independent Variables   
- Regression Line/Curve
- Residuals and Error :
    - Residual = Observed value - Predicted value
    - ما از جمع باقی مانده ها(حالا به هرروشی) میرسیم به خطا
<div dir='rtl'>

- Regression Line/Curve : در رگرسیون خطی ساده، رابطه میان متغیر مستقل (x) و متغیر وابسته (y) با استفاده از یک خط راست که «خط رگرسیون» نامیده می‌شود، مدل‌سازی می‌گردد. خط رگرسیون، خطی است که بهترین برازش(fit) را بر نقاط داده در یک نمودار پراکندگی (scatter plot) دارد و تفاوت میان مقادیر مشاهده‌شده و مقادیر پیش‌بینی‌شده توسط این خط را به حداقل می‌رساند.
</div>
