### **04-01**
## Regression
<div dir='rtl'>

**پیش‌بینی** مقدار یک متغیر بر اساس متغیر دیگر. مدل‌سازی رابطه‌ی بین یک متغیر پاسخ (وابسته) و یک یا چند متغیر پیش‌بین (مستقل) — معمولاً برای پیش‌بینی یا تبیین.

در اصل، رگرسیون به ما این امکان را می‌دهد که:

1. مقدار یک متغیر وابسته را بر اساس مقادیر متغیرهای مستقل *پیش‌بینی* کنیم.
2. ماهیت و شدت روابط میان متغیرها را *تبیین* (یا تشریح) کنیم.
</div>

**Types of Regression Problems** :
- Simple vs. Multiple Regression
    - $y = f(x) + \epsilon$
    - $y = f(x_1, x_2, ..., x_n) + \epsilon$
- Linear vs. Non-linear Regression
    - $y = f(X, \beta) + \epsilon$, where $f$ is a non-linear function
- Other Types of Regression
    1. **Logistic Regression**: Used for binary classification problems, but technically a regression method.
    2. **Ridge Regression**: Linear regression with L2 regularization to prevent overfitting.
    3. **Lasso Regression**: Linear regression with L1 regularization, useful for feature selection.
    4. **Elastic Net**: Combines both L1 and L2 regularization.
    5. **Polynomial Regression**: Fits a non-linear relationship using polynomial terms.
    6. **Stepwise Regression**: Iteratively adds or removes predictors to find the best model.

**Key Concepts in Regression** :  
- Dependent and Independent Variables   
- Regression Line/Curve
- Residuals and Error


<div dir='rtl'>

- Regression Line/Curve : در رگرسیون خطی ساده، رابطه میان متغیر مستقل (x) و متغیر وابسته (y) با استفاده از یک خط راست که «خط رگرسیون» نامیده می‌شود، مدل‌سازی می‌گردد. خط رگرسیون، خطی است که بهترین برازش(fit) را بر نقاط داده در یک نمودار پراکندگی (scatter plot) دارد و تفاوت میان مقادیر مشاهده‌شده و مقادیر پیش‌بینی‌شده توسط این خط را به حداقل می‌رساند.

- Residuals and Error :
  - Residual = Observed value - Predicted value
  - **Residual (باقیمانده):** برای **هر نقطه** جداگانه محاسبه می‌شه.
  
  - **Error (خطا):** یک مفهوم کلی‌تره. شامل خطای مدل، خطای اندازه‌گیری، و خطای ذاتی داده می‌شه.
  - **Loss Function (تابع زیان):** تابعی که **همه‌ی Residualها رو با هم** جمع می‌کنه تا یک عدد نهایی بده.
    - مثلاً MSE = میانگین مربع Residualها.
</div>

محور عمودی Residual هست :
- اگه Residual مثبت باشه → مدل کمتر از مقدار واقعی پیش‌بینی کرده. (نقطه بالای خط صفر)
- اگه Residual منفی باشه → مدل بیشتر از مقدار واقعی پیش‌بینی کرده. (نقطه زیر خط صفر)
- اگه Residual صفر باشه → مدل دقیقاً درست پیش‌بینی کرده. (نقطه روی خط صفر)


| جنبه | رگرسیون (Regression) | طبقه‌بندی (Classification) |
|--------|------------|----------------|
| **خروجی** | پیوسته (Continuous) | گسسته (Discrete) |
| **مثال خروجی** | قیمت، دما | دسته‌بندی، بله/خیر |
| **الگوریتم‌های رایج** | رگرسیون خطی، رگرسیون چندجمله‌ای | رگرسیون لجستیک، درخت تصمیم، SVM |
| **تابع زیان** | معمولاً از خطای مربعات(squared error) استفاده می‌کند | معمولاً از Log Loss یا Hinge Loss استفاده می‌کند |

مراحل آنالیز رگرسیون :(بیشتر مواقع مراحل خطی نیستند)

- ۳. پیش‌پردازش داده (Data Preprocessing)
- ۴. تحلیل اکتشافی (EDA) 
- ۵. انتخاب مدل (Model Selection)
- ۶. آموزش مدل (Training)
- ۷. ارزیابی مدل (Evaluation) ← R², MSE
- ۸. عیب‌یابی مدل (Diagnostics) ← Residual Plot
- ۹. اصلاح مدل (Refinement)
- ۱۰. تفسیر و گزارش (Interpretation)

### **04-02**
## Simple Linear Regression
<div dir='rtl'>

برای مدل‌سازی رابطه میان دو متغیر به کار می‌رود.

فرضیات در این مدل :
1. Linearity
2. Independence : استقلال مشاهدات(ردیف ها)
3. Homoscedasticity : واریانس Residualها در همه‌ی سطوح X ثابت باشه. یعنی پراکندگی خطاها با تغییر X تغییر نکنه.
4. Normality :
- خطا = y واقعی - y پیش‌بینی‌شده
- یعنی چی؟ یعنی اگه همه‌ی خطاها رو جمع کنی و نمودار توزیعشون رو بکشی، باید شبیه منحنی زنگوله‌ای (توزیع نرمال) باشه.
- به زبان ساده‌تر: خطاها باید تصادفی و متقارن دور صفر پخش شده باشن. بیشترشون نزدیک صفر، و هر چی دورتر می‌شیم، کمتر می‌شن.
5. No or Little Multicollinearity : متغیرهای مستقل (ستون‌های X)
- در مورد رگرسیون خطی ساده که تنها یک متغیر مستقل دارد، این فرض به‌طور خودکار برقرار است. با این حال، در رگرسیون خطی چندگانه، این فرض اهمیتی حیاتی پیدا می‌کند.
6. No Outliers
</div>

Checking Assumptions :
1. **Scatter plots**: To check linearity and homoscedasticity
2. **Q-Q plots**: To check normality of residuals. با هیسپلات هم میشه ولی چشمی میشه
3. **Residual plots**: To check linearity, homoscedasticity, and identify outliers
4. **Durbin-Watson test**: To check independence
