### **03-01**
## Optimization
<div dir='rtl'>

جنبه های کلیدی:

1. **Objective Function | تابع هدف**
2. **Variables or Parameters**: این‌ها عناصری هستند که می‌توانیم آن‌ها را برای بهینه‌سازی تابع هدف تنظیم کنیم.
- در مدل‌های یادگیری ماشین، این عناصر معمولاً همان وزن‌ها و بایاس‌های (bias) مدل هستند.
- هدف، یافتن مقادیر بهینه برای این متغیرهاست، به‌طوری که منجر به بهترین عملکرد مدل شود.
3. **Constraints | محدودیت ها**: شرایطی هستند که مقادیر ممکن متغیرها را محدود می‌کنند.
4. **loss function | تابع زیان**: برای سنجش میزان خطا میان مقادیر پیش‌بینی‌شده و مقادیر واقعی به کار می‌رود.
- انواعش مثل: Mean Squared Error (MSE) - Binary Cross-Entropy - Categorical Cross-Entropy

یکی از الگوریتم های رایج بهینه سازی Gradient Descent هست که خوش انواع مختلفی داره : Batch  - Stochastic (SGD) - Mini-Batch
