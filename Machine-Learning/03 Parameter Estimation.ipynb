### **03-01**
## Parameter Estimation in Machine Learning
<div dir='rtl'>

الان که ما اومدیم و پدیده های دنیای واقعی رو مدل کردیم(فرمش رو پیدا کردیم)(که ایشالاه درست باشه) حالا بیایم بپردازیم به جزئیات داخلش.

هدف اینکه بیام این پارامترهای داخل این فرم هارو به درستی تعیینشون بکنیم.

تعریف رسمی: وظیفه‌ی یافتن بهترین مجموعه پارامترها برای یک مدل، به‌گونه‌ای که توانایی آن را در تبیین یا پیش‌بینی داده‌های مشاهده‌شده به حداکثر برساند.
</div>

- Types of Parameters in Machine Learning Models   
  - Model Parameters : پارامترهایی که **در طول آموزش** یاد گرفته می‌شن. مثل وزن‌ها و بایاس‌های شبکه عصبی، یا ضرایب رگرسیون. 
  - Hyperparameters : قبل از شروع روند یادگیری تنظیم میشن
  - Latent Parameters    
  - Fixed Parameters    
  - Structural Parameters    
  - Regularization Parameters

- Parameter Estimation Methods  
  - Maximum Likelihood Estimation (MLE)
  - Bayesian Estimation   
  - Method of Moments (MoM)  
  - Maximum A Posteriori (MAP)  

<div dir='rtl'>

- MLE : جوری پارامترها رو پیداکن که Likelihood حداکثربشه.
> ⚠️ **نکته مهم:** Likelihood با Probability فرق داره!
> - **Probability:** پارامتر ثابت است، داده متغیر است. می‌پرسیم «احتمال دیدن داده چیست؟»
> - **Likelihood:** داده ثابت است (مشاهده شده)، پارامتر متغیر است. می‌پرسیم «با کدام پارامتر، این داده محتمل‌تر است؟»
> - در MLE، داده ثابت است و ما دنبال پارامتری می‌گردیم که Likelihood را حداکثر کند.
> - تعریف Likelihood با مثال: به شرطی که فرضیات من فلان چیز باشه(مثلا میانگین 4 و واریانس 2.5) چه قدر احتمال داره که من این مقدار(مثلا x=10) رو ببینم؟
- MAP : توی روش Posteriori، Bayesian Estimation رو حداکثر کنیم.
> - نکته : MAP = MLE + Prior. یعنی MAP یک نسخه‌ی بیزی از MLE است.
