### **02-04**
## Common Discrete Probability Distributions | توزیع احتمالاتی گسسته:
1. Bernoulli Distribution : (با یک بار آزمایش)دوحالت بیشتر نداره یا موفق یا شکست
2. Binomial Distribution : تعداد بار موفقیت یا شکست توی چندبار آزمایش مستقل
3. Multinomial Distribution : (تعدادبار حالت خاص توی چندبار آزمایش مستقل)باینومیالی که تعداد حالت های ممکنش بیشتر از دوتاست
4. Poisson Distribution : تعداد حالات تصادفی و مستقل که در یک بازه زمانی یا مکانی ثابت رخ می دهد(در این بازه چند رویداد؟)
5. Geometric Distribution : چندبار تکرار کنم(آزمایش های برنولی مستقل) تا به اولین موفقیتم برسم؟(صبر تا پیروز بشی)
6. Negative Binomial Distribution
7. Hypergeometric Distribution
<div dir='rtl'>

**Poisson Distribution** :

تنها یک پارامتر دارد. لامدا(λ) : نرخ متوسط وقوع _ میانگین تعداد رویدادها در بازه ثابت. میانگین = واریانس = لامدا

### **02-05**
## Common Continuous Probability Distributions | توزیع احتمالاتی پیوسته

1. Uniform Distribution : وقتی همه ی خروجی ها به یک انداره محتمل اند
2. Normal (Gaussian) Distribution
3. Exponential Distribution : (گذشته ربطی به آینده نداره)
4. Gamma Distribution : چقدر طول می‌کشه تا کاامین رویداد؟ 
5. Beta Distribution
<div dir='rtl'>

- **پواسون (گسسته):** «توی یه بازه، چند تا رویداد اتفاق میفته؟»
- **نمایی (پیوسته):** «چقدر طول می‌کشه تا رویداد بعدی؟»
- پواسون: «توی ۱ دقیقه، چند تا اتم فروپاشی می‌کنه؟»
- نمایی: «چقدر طول می‌کشه تا اتم بعدی فروپاشی کنه؟»
- تعمیم نمایی: نمایی = انتظار برای **اولین** رویداد؛ گاما = انتظار برای **kمین** رویداد در فرایند پواسون
