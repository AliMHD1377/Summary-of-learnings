قبل از استقرار، محیط های مختلفی ایجاد می کنیم:
1. Development :
    - همین لپ تاب خودمون یا میتونه یک سروری باشه
2. Testing :
    - مثل یک کامپیوتر خالی که از صفر بهش دستور میدیم تا بیاد بالا
3. Staging :
    - خیلی شبیه به خروجی نهایی.
    - توی مرحله تست فقط خودت کارمیکنی با حوزه کاریت(مثلا backend) ولی توی این مرحله همه هستند
4. Production :
    - میره که برسه دست user

- نکته : هرچی میریم جلو باید خرج بیشتری بکنیم برای همین خیلی از استارتاپ های کوچیک از 1 مستقیم میرن به 4
